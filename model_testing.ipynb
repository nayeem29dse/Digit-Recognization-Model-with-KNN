```python
import cv2
import pickle as pkl
import matplotlib.pyplot as plt
```

# Test Image -1

```python
test_img1=cv2.imread("sp1.png", cv2.IMREAD_GRAYSCALE)
test_img1
```

```python
test_img1.shape
```

```python
plt.imshow(test_img1, cmap='gray')
```

```python
test_img1M=cv2.bitwise_not(test_img1)
test_img1M
```

```python
plt.imshow(test_img1M, cmap='gray')
```

```python
test_img1M_resized = cv2.resize(test_img1M, (28, 28), interpolation=cv2.INTER_LINEAR)
test_img1M_resized
```

```python
plt.imshow(test_img1M_resized, cmap='gray')
```

```python
test_img1M_resized=test_img1M_resized/255
```

```python
test_img1M_resized_flat=test_img1M_resized.reshape(1, 28*28)
test_img1M_resized_flat
```

```python
with open('KNN_Model.pkl', 'rb') as file:
    model=pkl.load(file)
```

```python
model.predict(test_img1M_resized_flat)
```
